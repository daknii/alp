## Lista de exercícios - Algoritmos Iterativos

Este caderno reúne as **questões da seção Iterativos** da lista de problemas do WebAlgo, agora para resolução em **Python**.

### Orientações
- execute a célula de configuração abaixo uma vez;
- implemente cada solução dentro da função indicada;
- utilize `for` e/ou `while` de acordo com o problema;
- use estruturas condicionais quando forem necessárias;
- procure resolver cada problema antes de executar a validação;
- a saída deve conter apenas o que foi solicitado no exercício;
- quando houver mais de um valor de saída, escreva cada valor com `print()`.

````python
# @title
from unittest.mock import patch
from io import StringIO
import sys
import math

def _run(func, entradas):
    """Executa a função simulando valores informados por input()."""
    with patch("builtins.input", side_effect=[str(x) for x in entradas]):
        out = StringIO()
        old = sys.stdout
        sys.stdout = out
        try:
            func()
        finally:
            sys.stdout = old
    return out.getvalue().strip()

def _igual_item(recebido, esperado, tol=1e-6):
    try:
        return math.isclose(
            float(recebido),
            float(esperado),
            rel_tol=tol,
            abs_tol=tol
        )
    except (ValueError, TypeError):
        return str(recebido).strip() == str(esperado).strip()

def validar(func, testes, tol=1e-6):
    """Valida saídas simples ou múltiplos valores separados por espaços/linhas."""
    ok = 0

    for entradas, esperado in testes:
        try:
            saida = _run(func, entradas)

            if isinstance(esperado, (list, tuple)):
                recebidos = saida.split()
                esperados = [str(item) for item in esperado]

                passou = (
                    len(recebidos) == len(esperados)
                    and all(
                        _igual_item(recebido, item_esperado, tol)
                        for recebido, item_esperado in zip(recebidos, esperados)
                    )
                )
            else:
                passou = _igual_item(saida, esperado, tol)

            if passou:
                ok += 1
        except Exception:
            pass

    if ok == len(testes):
        print("✅ VALIDADO")
    else:
        print(f"❌ NÃO VALIDADO ({ok}/{len(testes)} testes)")
````

# Referência rápida — repetições em Python

## `for`

```python
for i in range(1, 6):
    print(i)
```

Produz:

```text
1
2
3
4
5
```

## `while`

```python
numero = int(input())

while numero >= 0:
    print(numero)
    numero = int(input())
```

## Contador

```python
contador = 0

for i in range(10):
    numero = int(input())
    if numero % 2 == 0:
        contador += 1
```

## Acumulador

```python
soma = 0

for i in range(10):
    numero = int(input())
    soma += numero
```

Lembre-se: em muitos problemas iterativos será necessário combinar **repetição + condição + variáveis de controle**.

````python
def ex20():
  menor = 11
  qtd = 0
  n = int(input("Digite um número:"))
  n1 = n
  while n > 0:
    N = int(input("Digite uma nota: "))
    if 0 <= N <= 10:
      if N == menor:
        qtd = qtd + 1

      elif menor > N:
        menor = N
        qtd = 1
      n = n-1
  porcentagem = (qtd/n1)*100
  print(menor)
  print(porcentagem)
````
