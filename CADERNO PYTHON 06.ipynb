**Validação:** os testes aceitam saídas numéricas em listas Python, em linhas ou separadas por espaços, inclusive com rótulos. Em exercícios com registros ou histogramas, siga a formatação indicada. Evite imprimir números extras, como numeração de itens ou mensagens de depuração.


```python
# @title Configuração e validação
from unittest.mock import patch
from io import StringIO
import sys
import math

def _run(func, entradas):
    """Executa a função com valores simulados em input()."""
    with patch("builtins.input", side_effect=[str(x) for x in entradas]):
        out = StringIO()
        old = sys.stdout
        sys.stdout = out
        try:
            func()
        finally:
            sys.stdout = old
    return out.getvalue().strip()

def _igual_item(recebido, esperado, tol=1e-6):
    try:
        return math.isclose(float(recebido), float(esperado), rel_tol=tol, abs_tol=tol)
    except (ValueError, TypeError):
        return str(recebido).strip() == str(esperado).strip()

def validar(func, testes, tol=1e-6):
    """Aceita números em linhas, separados por espaços ou em listas Python.

    Para resultados textuais (histogramas e registros), compara linhas.
    Textos descritivos antes de valores numéricos são permitidos.
    """
    import re
    padrao_numero = r"(?<![\w.])[-+]?(?:\d+(?:\.\d*)?|\.\d+)(?:[eE][-+]?\d+)?(?![\w.])"
    ok = 0
    for entradas, esperado in testes:
        try:
            saida = _run(func, entradas)
            if isinstance(esperado, (list, tuple)):
                # Uma saída com campos por linha exige a mesma divisão em linhas.
                if any(isinstance(v, str) for v in esperado):
                    recebidos = [linha.strip() for linha in saida.splitlines() if linha.strip()]
                    certo = len(recebidos) == len(esperado) and all(
                        _igual_item(a, b, tol) for a, b in zip(recebidos, esperado)
                    )
                else:
                    # Ignora rótulos, colchetes, vírgulas e espaços; preserva a ordem.
                    recebidos = re.findall(padrao_numero, saida)
                    certo = len(recebidos) == len(esperado) and all(
                        _igual_item(a, b, tol) for a, b in zip(recebidos, esperado)
                    )
            else:
                recebidos = re.findall(padrao_numero, saida) if isinstance(esperado, (int, float)) else None
                certo = (_igual_item(recebidos[0], esperado, tol) if recebidos is not None and len(recebidos) == 1
                         else _igual_item(saida, esperado, tol))
            if certo:
                ok += 1
            else:
                print(f"Falhou: entrada={entradas[:12]}{'...' if len(entradas)>12 else ''}; esperado={str(esperado)[:160]}; recebido={saida[:160]}")
        except Exception as erro:
            print(f"Erro: {type(erro).__name__}: {erro}")
    print(f"{ok}/{len(testes)} testes passaram")
```
